**2. What is an interpreter in Python?**

-> An interpreter in Python is a program that executes Python code line by line. It translates the
high-level Python code into machine-level instructions that the computer can understand. Unlike
compiled languages, where the entire program is converted into machine code before execution,
Python runs the code directly through the interpreter, making it easier to test and debug.

When we run a Python script (e.g., python script.py):
1. The interpreter first reads the code.
2. It parses it into bytecode (an intermediate, low-level representation).
3. Then the bytecode is executed by the Python virtual machine (PVM).

Example:

print("Hello, world!")

The interpreter:
* Reads this line.
* Converts it into bytecode.
* Executes it to display: Hello, world!

**3. What are pre-defined keywords in Python ?**

purposes in the language. These keywords are part of the Python syntax and cannot be used

Python code.

**4. Can keywords be used as variable names ?**

meanings in the language, using them as variable names will cause a syntax error. For
example, trying to do this:

will result in an error because if is a keyword used for conditional statements.

To avoid this, always choose variable names that are not part of Python’s reserved keywords. If you’re unsure which words are keywords, you can check them using:

This will display the full list of keywords that you should avoid using as variable names.

**6. Why are lists mutable, but tuples are immutable ?**

-> Lists are mutable because they are designed to allow modification after creation—elements can be added, removed, or changed. This flexibility makes lists useful for situations where we need a collection that can grow or change over time.

Tuples, on the other hand, are immutable because they are meant to represent fixed collections of items that shouldn’t change. Their immutability provides advantages like:
* Data integrity: Once created, the contents won’t accidentally be altered.
* Performance: Because tuples are immutable, Python can optimize their storage and access.
* Hashability: Tuples can be used as keys in dictionaries, while lists cannot, since keys must be immutable.

**7. What is the difference between “==” and “is” operators in Python?**

-> The difference between = = and is operators in Python is:
* == checks if the values of two objects are equal. It compares the contents or data inside the objects.
* is checks if two variables point to the exact same object in memory. It compares the identities (memory addresses) of the objects.

Example:

a = [1, 2, 3]

b = [1, 2, 3]

print(a == b)  # True, because their contents are the same

print(a is b)  # False, because they are two different objects in memory

c = a

print(a is c)  # True, because both refer to the same object

So, use = = when you want to check if two objects have the same value, and use is when you want to
check if they are actually the same object.

**9. What is type casting in Python?**

-> Type casting in Python refers to the process of converting a value from one data type to another, either automatically or manually. Python performs implicit type casting when it
automatically converts one data type to another during operations—for example, adding an integer and a float will result in a float. Explicit type casting, on the other hand, is when the
programmer manually changes the data type using functions like int(), float(), or str(). This is useful when you're working with user input (which is often a string) and need it in a numeric form for calculations.

  Example:
  
  x = "5"
  
  y = int(x)  # converting string to integer
  
  print(y + 3)  # Output: 8

**10. What is the difference between implicit and explicit type casting?**

-> Implicit type casting in Python is when the interpreter automatically converts one data type to another during an operation. This usually happens when combining different types in an expression, and Python safely promotes the value to a more general or compatible type. For example, if you add an integer and a float, Python will automatically convert the integer to a float to avoid losing information.

  Implicit Casting:

  x = 5       # int

  y = 2.0     # float

  z = x + y   # Python automatically converts x to float

  print(z)    # Output: 7.0

Explicit type casting, on the other hand, is when the programmer manually converts a value from one data type to another using functions like int(), float(), str(), or bool(). This is useful
when dealing with data types that Python cannot convert automatically, such as converting a string input like "10" into an integer for mathematical operations. Unlike implicit casting,
explicit casting requires care, as trying to convert incompatible types (like "abc" to int) can lead to errors.

   Explicit Casting:

  x = "10"     # string   

  y = int(x)   # manually cast to int

  print(y + 5) # Output: 15

**12. How does the elif statement work?**

->  The **elif** statement in Python stands for "else if" and is used to check multiple conditions in a sequence. It allows you to test more than two possibilities without writing multiple if or nested if-else statements.

*How It Works:*
* Python evaluates the if condition first.
* If the if condition is False, it moves to the elif condition.
* If an elif condition is True, its block of code runs, and the rest are skipped.
* If none of the if or elif conditions are true, the else block (if present) runs.

  Example:

  marks = 75

  if marks >= 90:
  
    print("Grade: A")
  
  elif marks >= 75:

    print("Grade: B")

  elif marks >= 60:

    print("Grade: C")
  
  else:
    
    print("Grade: D")
  
  Output:
  
  Grade: B

**13. What is the difference between for and while loops?**

-> The for loop in Python is used when you know exactly how many times you want to repeat a block of code. It's commonly used to iterate over sequences like lists, strings, or ranges. The loop runs once for each item in the sequence and stops automatically when the
sequence ends.

  Example:

  for i in range(5):

    print(i)

This will print numbers from 0 to 4. The loop runs five times because range(5) generates the numbers 0 through 4.

The while loop is used when you want to repeat a block of code as long as a condition remains true. It's useful when the number of repetitions is not known in advance. You must manually update the condition inside the loop, or it may result in an infinite loop.

  Example:
  
  i = 0

  while i < 5:

    print(i)
    
    i += 1

This loop also prints numbers from 0 to 4. It continues running as long as the condition i < 5 is true, and stops when i reaches 5.

**14. Describe a scenario where a while loop is more suitable than a for loop?**

-> A scenario where a while loop is more suitable than a for loop is when you need to repeat an action an unknown number of times until a specific condition is met. For example, consider a password verification system where a user is asked to enter the correct
password. Since you don't know how many attempts the user will need, a while loop is ideal because it allows the program to keep running as long as the entered password is incorrect.

  Example:

  correct_password = "python123"
  
  user_input = ""

  while user_input != correct_password:

    user_input = input("Enter the password: ")

  print("Access granted!")

In this case, the while loop keeps prompting the user until they type the correct password. A for loop wouldn't be appropriate here because it requires a fixed number of iterations, which is not suitable when the number of tries is unpredictable.

```python
#  14. Write a program to reverse a string using a while loop.

text = input("Enter a string: ")

reversed_text = ""
index = len(text) - 1

while index >= 0:
    reversed_text += text[index]
    index -= 1

print("Reversed string:", reversed_text)
```

```python
#  15. Write a Python program to calculate the factorial of a number provided by the user using a while loop.

num = int(input("Enter a non-negative integer: "))

factorial = 1
i = 1

while i <= num:
    factorial *= i
    i += 1

print(f"Factorial of {num} is {factorial}")
```
